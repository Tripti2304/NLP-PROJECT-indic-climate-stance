```python
!pip install -q huggingface_hub openpyxl pyarrow

import os, glob
import pandas as pd
from huggingface_hub import snapshot_download

path = snapshot_download(repo_id="ciol-research/Dhoroni", repo_type="dataset")
files = [f for f in glob.glob(path + "/**/*", recursive=True)
         if os.path.isfile(f) and not f.lower().endswith((".md", ".gitattributes"))]
print("Files in the dataset repo:")
for f in files:
    print(" ", f.replace(path, ""), "-", round(os.path.getsize(f) / 1024), "KB")

def read_any(f):
    e = f.lower().rsplit(".", 1)[-1]
    if e == "csv":              return pd.read_csv(f, keep_default_na=False)
    if e == "tsv":              return pd.read_csv(f, sep="\t", keep_default_na=False)
    if e in ("xlsx", "xls"):    return pd.read_excel(f)
    if e == "json":             return pd.read_json(f)
    if e == "jsonl":            return pd.read_json(f, lines=True)
    if e == "parquet":          return pd.read_parquet(f)
    return None

frames = {}
for f in files:
    try:
        d = read_any(f)
        if d is not None:
            frames[f] = d
            print("\nREAD:", f.replace(path, ""), d.shape)
    except Exception as ex:
        print("could not read", f.replace(path, ""), "->", ex)

# use the biggest table as the main dataframe
main_file = max(frames, key=lambda k: len(frames[k]))
df = frames[main_file]
print("\nUsing:", main_file.replace(path, ""))
print(df.shape)
print(df.columns.tolist())
df.head(3)
```

```python
for c in df.columns:
    n = df[c].astype(str).nunique()
    if n <= 15:
        print("\n==", c, "==")
        print(df[c].astype(str).value_counts())
```

```python
# 1) Text column = the text column with the longest average length (news articles)
obj_cols = [c for c in df.columns if df[c].dtype == object]
TEXT_COL = max(obj_cols, key=lambda c: df[c].astype(str).str.len().mean())

# 2) Stance column = a column whose name contains "stance"
cands = [c for c in df.columns if "stance" in str(c).lower()]
if not cands:
    print("No column named 'stance'. Columns with few unique values (candidates):")
    for c in df.columns:
        if df[c].astype(str).nunique() <= 10:
            print(f"\n== {c} ==")
            print(df[c].astype(str).value_counts())
    raise SystemExit("Pick the stance column from the list above and set STANCE_COL by hand.")
STANCE_COL = cands[0]

print("TEXT_COL   =", TEXT_COL)
print("STANCE_COL =", STANCE_COL)
print("\nStance values found:")
print(df[STANCE_COL].astype(str).str.strip().value_counts())

# 3) Auto-map the values to Favour / Against / None
def auto_map(v):
    s = str(v).strip().lower()
    if any(k in s for k in ["neutral", "none", "unrelated", "no stance", "unclear", "not applicable", "irrelevant", "unknown"]):
        return "None"
    if any(k in s for k in ["against", "oppos", "deny", "denial", "disagree", "negative", "skeptic", "dismiss"]):
        return "Against"
    if any(k in s for k in ["favor", "favour", "support", "agree", "positive", "pro"]):
        return "Favour"
    return None

values = df[STANCE_COL].astype(str).str.strip().str.lower().unique()
LABEL_MAP = {v: auto_map(v) for v in values if auto_map(v) is not None}

# 4) Manual fixes: add or correct anything the auto-map got wrong or missed.
#    Example: MANUAL = {"0": "Against", "1": "None", "2": "Favour"}
MANUAL = {}
LABEL_MAP.update({k.lower(): v for k, v in MANUAL.items()})

print("\nMapping that will be used:")
for k in values:
    print(f"  {k!r:35} -> {LABEL_MAP.get(k, 'DROPPED (unmapped)')}")
```

```python
bn = pd.DataFrame({
    "text": df[TEXT_COL].astype(str),
    "label": df[STANCE_COL].astype(str).str.strip().str.lower()
                .map({k.lower(): v for k, v in LABEL_MAP.items()}),
    "language": "bn",
    "source": "Dhoroni",
})
bn = bn.dropna(subset=["label"])
print(len(bn), "Bengali rows kept")
print(bn["label"].value_counts())
```

```python
hi = pd.read_csv("hindi.csv", keep_default_na=False)   # keep_default_na is required for the "None" label
hi["language"] = "hi"
hi["source"] = "self-written"
data = pd.concat([bn, hi[["text","label","language","source","style"]]],
                 ignore_index=True)
data["text"] = data["text"].str.strip()
data = data[data["text"] != ""].drop_duplicates("text").reset_index(drop=True)
print(len(data))
print(pd.crosstab(data["language"], data["label"]))
```

```python
data["n_words"] = data["text"].str.split().str.len()

def latin_ratio(t):
    letters = re.findall(r"[^\W\d_]", t)
    return sum(c.isascii() for c in letters) / max(len(letters), 1)
data["latin_ratio"] = data["text"].apply(latin_ratio)

print("Instances:", len(data))
print("\nBy language:\n", data["language"].value_counts())
print("\nClass distribution:\n", data["label"].value_counts())
print("\nWords per text:\n", data["n_words"].describe().round(1))
print("\nCode-mixed/Roman-script share (>20% Latin letters):",
      round((data["latin_ratio"] > 0.2).mean()*100, 1), "%")
print("Texts with emoji:", data["text"].str.contains(r"[\U0001F300-\U0001FAFF]").sum())
print("Texts with hashtags:", data["text"].str.contains("#").sum())
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 4))
data["label"].value_counts().plot.bar(ax=ax[0], title="Class distribution")
pd.crosstab(data["language"], data["label"]).plot.bar(ax=ax[1], title="Class by language")
data["n_words"].clip(upper=data["n_words"].quantile(0.99)).plot.hist(bins=40, ax=ax[2], title="Words per text")
plt.tight_layout(); plt.savefig("dataset_stats.png", dpi=150); plt.show()
```

```python
from sklearn.model_selection import train_test_split

d = data.copy()
d["strat"] = d["language"] + "_" + d["label"]
d.loc[d["strat"].map(d["strat"].value_counts()) < 6, "strat"] = "rare"

train, temp = train_test_split(d, test_size=0.30, random_state=42, stratify=d["strat"])
try:
    val, test = train_test_split(temp, test_size=0.50, random_state=42, stratify=temp["strat"])
except ValueError:
    val, test = train_test_split(temp, test_size=0.50, random_state=42)

for name, part in [("train", train), ("val", val), ("test", test)]:
    part.drop(columns=["strat"]).reset_index(drop=True).to_csv(f"{name}.csv", index=False)
    print(name, len(part))
data.to_csv("climate_stance_dataset.csv", index=False)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline, make_union
from sklearn.metrics import classification_report, f1_score, ConfusionMatrixDisplay

tr = pd.read_csv("train.csv", keep_default_na=False)
te = pd.read_csv("test.csv", keep_default_na=False)

vec = make_union(
    TfidfVectorizer(ngram_range=(1,2), min_df=2, sublinear_tf=True),
    TfidfVectorizer(analyzer="char_wb", ngram_range=(2,5), min_df=2, sublinear_tf=True),
)
clf = make_pipeline(vec, LogisticRegression(max_iter=2000, class_weight="balanced"))
clf.fit(tr["text"], tr["label"])
pred = clf.predict(te["text"])

print("BASELINE Macro F1:", round(f1_score(te["label"], pred, average="macro"), 4))
print(classification_report(te["label"], pred, digits=3))
for lang, g in te.groupby("language"):
    print(lang, "Macro F1:", round(f1_score(g["label"], clf.predict(g["text"]), average="macro"), 4))
ConfusionMatrixDisplay.from_predictions(te["label"], pred, xticks_rotation=45)
plt.tight_layout(); plt.savefig("baseline_confusion.png", dpi=150); plt.show()
```

```python
!pip install -q transformers accelerate scikit-learn pandas matplotlib

import os, re, time, random
import numpy as np, pandas as pd, torch
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader
from transformers import AutoTokenizer, AutoModelForSequenceClassification, get_linear_schedule_with_warmup
from sklearn.metrics import f1_score, accuracy_score, classification_report, ConfusionMatrixDisplay

SEED, MAXLEN, BS, LR, EPOCHS = 42, 128, 16, 2e-5, 4    # EPOCHS=2 if short on time; BS=8 if out of memory
MODELS = {"muril": "google/muril-base-cased", "xlmr": "xlm-roberta-base"}
LABELS = ["Against", "Favour", "None"]
L2I = {l: i for i, l in enumerate(LABELS)}
dev = "cuda" if torch.cuda.is_available() else "cpu"
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
os.makedirs("results", exist_ok=True)
print("device:", dev)
```

```python
def load(p):
    d = pd.read_csv(p, keep_default_na=False)      # required so the label "None" stays a label
    return d[d["label"].isin(LABELS)].reset_index(drop=True)

train, val, test = load("train.csv"), load("val.csv"), load("test.csv")
print(len(train), len(val), len(test))
print(train["label"].value_counts())
print(test.groupby("language")["label"].value_counts())
```

```python
def make_loader(df, tok, shuffle):
    texts, ys = df["text"].tolist(), df["label"].map(L2I).tolist()
    def collate(idx):
        enc = tok([texts[i] for i in idx], truncation=True, max_length=MAXLEN,
                  padding=True, return_tensors="pt")
        enc["labels"] = torch.tensor([ys[i] for i in idx])
        return enc
    return DataLoader(list(range(len(texts))), batch_size=BS, shuffle=shuffle, collate_fn=collate)

@torch.no_grad()
def predict_probs(model, loader):
    model.eval(); out = []
    for b in loader:
        b = {k: v.to(dev) for k, v in b.items() if k != "labels"}
        out.append(torch.softmax(model(**b).logits, -1).cpu().numpy())
    return np.concatenate(out)

def macro(df, probs):
    return f1_score(df["label"].map(L2I), probs.argmax(1), average="macro")

def train_model(name, ckpt):
    t0 = time.time()
    tok = AutoTokenizer.from_pretrained(ckpt)
    model = AutoModelForSequenceClassification.from_pretrained(ckpt, num_labels=3).to(dev)
    tr_l, va_l, te_l = make_loader(train, tok, True), make_loader(val, tok, False), make_loader(test, tok, False)
    cnt = train["label"].map(L2I).value_counts().reindex(range(3)).fillna(1).values
    w = torch.tensor(len(train) / (3 * cnt), dtype=torch.float).to(dev)    # class weights
    lossf = torch.nn.CrossEntropyLoss(weight=w)
    opt = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=0.01)
    steps = EPOCHS * len(tr_l)
    sch = get_linear_schedule_with_warmup(opt, int(0.1 * steps), steps)
    scaler = torch.cuda.amp.GradScaler(enabled=(dev == "cuda"))
    best, best_state = -1, None
    for ep in range(EPOCHS):
        model.train(); tot = 0
        for b in tr_l:
            b = {k: v.to(dev) for k, v in b.items()}
            y = b.pop("labels")
            with torch.autocast(device_type=dev, enabled=(dev == "cuda")):
                loss = lossf(model(**b).logits.float(), y)
            opt.zero_grad(); scaler.scale(loss).backward()
            scaler.unscale_(opt); torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            scaler.step(opt); scaler.update(); sch.step(); tot += loss.item()
        vf = macro(val, predict_probs(model, va_l))
        print(f"{name} epoch {ep+1}/{EPOCHS} | loss {tot/len(tr_l):.4f} | val Macro F1 {vf:.4f}")
        if vf > best:        # best epoch chosen on VALIDATION only, never on test
            best = vf
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
    model.load_state_dict(best_state)
    probs = predict_probs(model, te_l)
    np.save(f"results/probs_{name}.npy", probs)
    del model; torch.cuda.empty_cache()
    return probs, time.time() - t0

all_probs, times = {}, {}
print("helpers ready")
```

```python
all_probs["muril"], times["muril"] = train_model("muril", MODELS["muril"])
print("MuRIL test Macro F1:", round(macro(test, all_probs["muril"]), 4))
```

```python
all_probs["xlmr"], times["xlmr"] = train_model("xlmr", MODELS["xlmr"])
print("XLM-R test Macro F1:", round(macro(test, all_probs["xlmr"]), 4))
```

```python
if len(all_probs) > 1:
    all_probs["ensemble"] = np.mean([all_probs[k] for k in list(all_probs)], axis=0)
    times["ensemble"] = 0

y = test["label"].map(L2I).values
rows = []
for name, p in all_probs.items():
    pred = p.argmax(1)
    r = {"model": name, "macro_f1": f1_score(y, pred, average="macro"),
         "accuracy": accuracy_score(y, pred), "train_time_sec": round(times[name])}
    for i, l in enumerate(LABELS):
        r[f"f1_{l}"] = f1_score(y, pred, labels=[i], average="macro")
    for lang, g in test.groupby("language"):
        r[f"macro_f1_{lang}"] = f1_score(y[g.index], pred[g.index], average="macro")
    rows.append(r)
    print("\n====", name, "====")
    print(classification_report(y, pred, target_names=LABELS, digits=3))
    ConfusionMatrixDisplay.from_predictions(y, pred, display_labels=LABELS, xticks_rotation=45)
    plt.title(name); plt.tight_layout()
    plt.savefig(f"results/confusion_{name}.png", dpi=150); plt.show()

res = pd.DataFrame(rows).round(4)
res.to_csv("results/results.csv", index=False)
res
```

```python
best_name = res.sort_values("macro_f1", ascending=False).iloc[0]["model"]
print("Best model:", best_name)
test = test.copy()
test["pred"] = [LABELS[i] for i in all_probs[best_name].argmax(1)]
NEG = r"नहीं|\bन\b|मत|nahi|nahin|\bna\b|\bnot\b|\bno\b|না|নেই|নয়"
test["has_negation"] = test["text"].str.contains(NEG, case=False, regex=True)

def latin_ratio(t):
    l = re.findall(r"[^\W\d_]", t)
    return sum(c.isascii() for c in l) / max(len(l), 1)
test["code_mixed"] = test["text"].apply(latin_ratio) > 0.2

for col in ["has_negation", "code_mixed", "style"]:
    if col not in test: continue
    for v, g in test.groupby(col):
        if v != "" and len(g) >= 5:
            print(f"{col} = {v}: n={len(g)}  Macro F1 = {f1_score(g['label'], g['pred'], average='macro'):.3f}")
```

```python
test["confidence"] = all_probs[best_name].max(1)
err = test[test["label"] != test["pred"]].sort_values("confidence", ascending=False)
err[["text", "label", "pred", "confidence", "language"]].head(30).to_csv("results/errors.csv", index=False)
print(len(err), "errors out of", len(test))
print(pd.crosstab(err["label"], err["pred"]))
err[["text", "label", "pred", "confidence"]].head(10)
```
